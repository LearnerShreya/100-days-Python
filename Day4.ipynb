#### Random Module
- integer : random.randint(1,11)
- float : random.random() , random.random() * 5

```python
import random

random_integer = random.randint(1,11)
print(random_integer)
```

```python
import random

# range(0,1)
random_float = random.random()
print(random_float)
```

```python
import random

# range(0,5)
random_float = random.random() * 5
print(random_float)
```

```python
choose = int(input("Choose 0 or 1: "))
import random
computer = random.randint(0,1)

if choose == 1 and computer == 1:
    print("Heads\nor\nHeads")
elif choose == 0 and computer == 1:
    print("Heads\nor\nHeads")
elif choose == 1 and computer == 0:
    print("Heads\nor\nTails")
else:
    print("Tails\nor\nTails")
```
