## @ Python Primitive Data Types

### Complete List of Data Types in Python

#### Numeric Types
- **int**: Represents integer values. These are whole numbers without a decimal point.
  ```python
  integer = 10
  ```

- **float**: Represents floating-point (decimal) values. These are numbers with a decimal point.
  ```python
  floating_point = 10.5
  ```

- **complex**: Represents complex numbers, which have a real part and an imaginary part.
  ```python
  complex_number = 3 + 4j
  ```

#### Sequence Types
- **str**: Represents string values, which are sequences of characters. Strings are immutable.
  ```python
  string = "Hello, Shreya!"
  ```

- **list**: Represents ordered collections of items, which can be of mixed data types. Lists are mutable.
  ```python
  list_example = [1, 2, 3, 4, 5]
  ```

- **tuple**: Represents ordered collections of items, which can be of mixed data types. Tuples are immutable.
  ```python
  tuple_example = (1, 2, 3, 4, 5)
  ```

- **range**: Represents an immutable sequence of numbers, typically used for looping a specific number of times.
  ```python
  range_example = range(1, 6)
  ```

#### Set Types
- **set**: Represents unordered collections of unique items. Sets are mutable.
  ```python
  set_example = {1, 2, 3, 4, 5}
  ```

- **frozenset**: Represents immutable sets.
  ```python
  frozenset_example = frozenset([1, 2, 3, 4, 5])
  ```

#### Mapping Type
- **dict**: Represents unordered collections of key-value pairs, where keys are unique. Dictionaries are mutable.
  ```python
  dict_example = {'name': 'Shreya', 'age': 21, 'city': 'Banka'}
  ```

#### Boolean Type
- **bool**: Represents Boolean values, which can be either `True` or `False`.
  ```python
  is_true = True
  is_false = False
  ```

#### Binary Types
- **bytes**: Represents immutable sequences of bytes.
  ```python
  byte_example = bytes(b'hello')
  ```

- **bytearray**: Represents mutable sequences of bytes.
  ```python
  byte_array = bytearray(b'hello')
  ```

- **memoryview**: Provides a view of the memory of an object without copying it. Useful for handling large data buffers.
  ```python
  memory_view = memoryview(bytes(5))
  ```

#### None Type
- **NoneType**: Represents the absence of a value or a null value. It is often used to signify 'nothing' or 'no value here'.
  ```python
  none_example = None
  ```

````python
# Numeric Types
integer = 10                                         # Integer
floating_point = 10.5                                # Float
complex_number = 3 + 4j                              # Complex

# Sequence Types
string = "Hello, Shreya!"                            # String
list_example = [1, 2, 3, 4, 5]                       # List
tuple_example = (1, 2, 3, 4, 5)                      # Tuple
range_example = range(1, 6)                          # Range

# Set Types
set_example = {1, 2, 3, 4, 5}                        # Set
frozenset_example = frozenset([1, 2, 3, 4, 5])       # Frozenset

# Mapping Type
dict_example = {'name': 'Shreya', 'age': 21, 'city': 'Banka'}  # Dictionary

# Boolean Type
is_true = True                                       # Boolean True
is_false = False                                     # Boolean False

# Binary Types
memory_view = memoryview(bytes(5))                   # Memoryview
byte_array = bytearray(b'hello')                     # Bytearray
byte_example = bytes(b'hello')                       # Bytes

# None Type
none_example = None                                  # NoneType

# Printing the values
print(f'Integer: {integer}')
print(f'Float: {floating_point}')
print(f'Complex: {complex_number}')

print(f'\nString: {string}')
print(f'List: {list_example}')
print(f'Tuple: {tuple_example}')
print(f'Range: {list(range_example)}')

print(f'\nSet: {set_example}')
print(f'Frozenset: {frozenset_example}')

print(f'\nDictionary: {dict_example}')

print(f'\nBoolean True: {is_true}')
print(f'Boolean False: {is_false}')

print(f'\nMemoryview: {memory_view}')
print(f'Bytearray: {byte_array}')
print(f'Bytes: {byte_example}')

print(f'\nNoneType: {none_example}')
````

### String Operations

1. **Concatenation**
   - Joining two or more strings together using the `+` operator.
   ```python
   str1 = "Hello"
   str2 = "World"
   result = str1 + " " + str2  # "Hello World"
   ```

2. **Repetition**
   - Repeating a string multiple times using the `*` operator.
   ```python
   result = "Hello" * 3  # "HelloHelloHello"
   ```

3. **Indexing**
   - Accessing individual characters in a string using their index.
   ```python
   string = "Hello"
   first_char = string[0]  # 'H'
   last_char = string[-1]  # 'o'
   ```

4. **Slicing**
   - Extracting a substring using a range of indices.
   ```python
   string = "Hello World"
   substring = string[0:5]  # "Hello"
   substring = string[6:]   # "World"
   substring = string[:5]   # "Hello"
   substring = string[-5:]  # "World"
   ```

5. **Length**
   - Getting the length of a string using the `len()` function.
   ```python
   string = "Hello"
   length = len(string)  # 5
   ```

### String Methods

1. **Uppercase and Lowercase**
   - Converting all characters in a string to uppercase or lowercase.
   ```python
   string = "Hello World"
   upper_string = string.upper()  # "HELLO WORLD"
   lower_string = string.lower()  # "hello world"
   ```

2. **Strip**
   - Removing leading and trailing whitespace (or specified characters).
   ```python
   string = "   Hello World   "
   stripped_string = string.strip()  # "Hello World"
   ```

3. **Replace**
   - Replacing a substring with another substring.
   ```python
   string = "Hello World"
   new_string = string.replace("World", "Python")  # "Hello Python"
   ```

4. **Split and Join**
   - Splitting a string into a list of substrings based on a delimiter and joining a list of strings into a single string.
   ```python
   string = "Hello World"
   words = string.split(" ")  # ["Hello", "World"]
   new_string = " ".join(words)  # "Hello World"
   ```

5. **Find and Count**
   - Finding the index of the first occurrence of a substring and counting the number of occurrences of a substring.
   ```python
   string = "Hello World"
   index = string.find("World")  # 6
   count = string.count("o")  # 2
   ```

6. **Startswith and Endswith**
   - Checking if a string starts or ends with a specified substring.
   ```python
   string = "Hello World"
   starts = string.startswith("Hello")  # True
   ends = string.endswith("World")  # True
   ```

### String Immutability
Strings in Python are immutable, which means you cannot modify them in place. Instead, you create new strings with the desired changes. This implies that there are no push or pop operations directly applicable to strings, as these typically involve modifying the sequence in place.

````python
# Concatenation
str1 = "Hello"
str2 = "World"
result = str1 + " " + str2
print(f'Concatenation: {result}')  # "Hello World"

# Repetition
result = "Hello" * 3
print(f'Repetition: {result}')  # "HelloHelloHello"

# Indexing
string = "Hello"
first_char = string[0]
last_char = string[-1]
print(f'First character: {first_char}')  # 'H'
print(f'Last character: {last_char}')  # 'o'

# Slicing
string = "Hello World"
substring1 = string[0:5]
substring2 = string[6:]
substring3 = string[:5]
substring4 = string[-5:]
print(f'Slicing: {substring1}, {substring2}, {substring3}, {substring4}')  # "Hello", "World", "Hello", "World"

# Length
length = len(string)
print(f'Length: {length}')  # 11

# Uppercase and Lowercase
upper_string = string.upper()
lower_string = string.lower()
print(f'Uppercase: {upper_string}')  # "HELLO WORLD"
print(f'Lowercase: {lower_string}')  # "hello world"

# Strip
stripped_string = string.strip()
print(f'Stripped: {stripped_string}')  # "Hello World"

# Replace
new_string = string.replace("World", "Python")
print(f'Replace: {new_string}')  # "Hello Python"

# Split and Join
words = string.split(" ")
new_string = " ".join(words)
print(f'Split: {words}')  # ["Hello", "World"]
print(f'Join: {new_string}')  # "Hello World"

# Find and Count
index = string.find("World")
count = string.count("o")
print(f'Find: {index}')  # 6
print(f'Count: {count}')  # 2

# Startswith and Endswith
starts = string.startswith("Hello")
ends = string.endswith("World")
print(f'Startswith: {starts}')  # True
print(f'Endswith: {ends}')  # True
````

## @ Mathematical Operations in Python

PEMDSAS LR ( Left to Right )

1. **Parentheses (P)**:
   - First, perform operations inside parentheses `()`.
   - Example: In `2 * (3 + 4)`, calculate `3 + 4` first to get `2 * 7 = 14`.

2. **Exponents (E)**:
   - Next, evaluate exponents and roots.
   - Example: In `3^2 + 4`, calculate `3^2` first to get `9 + 4 = 13`.

3. **Multiplication and Division (MD)**:
   - Then, perform multiplication and division from left to right.
   - Example: In `8 / 4 * 2`, perform `8 / 4` first to get `2 * 2 = 4`.

4. **Addition and Subtraction (AS)**:
   - Finally, perform addition and subtraction from left to right.
   - Example: In `7 - 3 + 2`, perform `7 - 3` first to get `4 + 2 = 6`.

### Python Operators

#### 1. Arithmetic Operators:
- **Addition (`+`)**: Adds two operands.
  ```python
  5 + 3  # 8
  ```
- **Subtraction (`-`)**: Subtracts the second operand from the first.
  ```python
  5 - 3  # 2
  ```
- **Multiplication (`*`)**: Multiplies two operands.
  ```python
  5 * 3  # 15
  ```
- **Division (`/`)**: Divides the first operand by the second (result is float).
  ```python
  5 / 3  # 1.6666666666666667
  ```
- **Floor Division (`//`)**: Divides and floors the result to the nearest integer.
  ```python
  5 // 3  # 1
  ```
- **Modulus (`%`)**: Returns the remainder of the division.
  ```python
  5 % 3  # 2
  ```
- **Exponentiation (`**`)**: Raises the first operand to the power of the second.
  ```python
  5 ** 3  # 125
  ```

#### 2. Comparison Operators:
- **Equal (`==`)**: Checks if two operands are equal.
  ```python
  5 == 3  # False
  ```
- **Not Equal (`!=`)**: Checks if two operands are not equal.
  ```python
  5 != 3  # True
  ```
- **Greater Than (`>`)**: Checks if the first operand is greater than the second.
  ```python
  5 > 3  # True
  ```
- **Less Than (`<`)**: Checks if the first operand is less than the second.
  ```python
  5 < 3  # False
  ```
- **Greater Than or Equal To (`>=`)**: Checks if the first operand is greater than or equal to the second.
  ```python
  5 >= 3  # True
  ```
- **Less Than or Equal To (`<=`)**: Checks if the first operand is less than or equal to the second.
  ```python
  5 <= 3  # False
  ```

#### 3. Assignment Operators:
- **Assign (`=`)**: Assigns the value of the right operand to the left operand.
  ```python
  a = 5
  ```
- **Add and Assign (`+=`)**: Adds and assigns the result.
  ```python
  a += 3  # a is now 8
  ```
- **Subtract and Assign (`-=`)**: Subtracts and assigns the result.
  ```python
  a -= 3  # a is now 2
  ```
- **Multiply and Assign (`*=`)**: Multiplies and assigns the result.
  ```python
  a *= 3  # a is now 15
  ```
- **Divide and Assign (`/=`)**: Divides and assigns the result.
  ```python
  a /= 3  # a is now 1.6666666666666667
  ```
- **Floor Divide and Assign (`//=`)**: Floor divides and assigns the result.
  ```python
  a //= 3  # a is now 1
  ```
- **Modulus and Assign (`%=`)**: Modulus and assigns the result.
  ```python
  a %= 3  # a is now 2
  ```
- **Exponentiate and Assign (`**=`)**: Exponentiates and assigns the result.
  ```python
  a **= 3  # a is now 125
  ```

#### 4. Logical Operators:
- **And (`and`)**: True if both operands are true.
  ```python
  True and False  # False
  ```
- **Or (`or`)**: True if at least one operand is true.
  ```python
  True or False  # True
  ```
- **Not (`not`)**: True if the operand is false.
  ```python
  not True  # False
  ```

#### 5. Bitwise Operators:
- **And (`&`)**: Bitwise AND.
  ```python
  5 & 3  # 1
  ```
- **Or (`|`)**: Bitwise OR.
  ```python
  5 | 3  # 7
  ```
- **Xor (`^`)**: Bitwise XOR.
  ```python
  5 ^ 3  # 6
  ```
- **Not (`~`)**: Bitwise NOT.
  ```python
  ~5  # -6
  ```
- **Left Shift (`<<`)**: Shifts bits left.
  ```python
  5 << 1  # 10
  ```
- **Right Shift (`>>`)**: Shifts bits right.
  ```python
  5 >> 1  # 2
  ```

#### 6. Membership Operators:
- **In (`in`)**: True if value is in the specified sequence.
  ```python
  'a' in 'apple'  # True
  ```
- **Not In (`not in`)**: True if value is not in the specified sequence.
  ```python
  'b' not in 'apple'  # True
  ```

#### 7. Identity Operators:
- **Is (`is`)**: True if variables point to the same object.
  ```python
  a is b
  ```
- **Is Not (`is not`)**: True if variables do not point to the same object.
  ```python
  a is not b
  ```

These operators cover a wide range of functionalities, from basic arithmetic to complex logical operations, providing the necessary tools for programming in Python.
