**indentation**-
In Python, indentation is not just for readability - it's a crucial part of the syntax. It defines blocks of code, such as loops, functions, and conditional statements.
Here's what you need to know about indentation in Python:
- Importance:
Indentation specifies which statements belong to which block. Unlike many other languages that use curly braces {} to define blocks, Python relies solely on indentation.

Rules:
- Consistent Spacing: Use the same number of spaces or tabs for each indentation level within a block.
- Common Practice: The standard is to use 4 spaces per indentation level.
- Errors: Incorrect indentation will raise an IndentationError.

### Types of Comments in Python

1. **Single-line Comments**
   - Start with `#` and extend to the end of the line.
   ```python
   # This is a single-line comment
   x = 5  # Inline comment
   ```

2. **Multi-line Comments**
   - Use multiple `#` characters or triple-quoted strings (`"""` or `'''`).
   ```python
   # This is a multi-line comment
   # that spans multiple lines.

   """
   This is a multi-line comment
   using triple-quoted strings.
   """
   ```

3. **Docstrings**
   - Used to document functions, classes, or modules. Written using triple-quoted strings and accessible via the `__doc__` attribute.
   ```python
   def example_function():
       """
       This is a docstring.
       It explains the purpose of the function.
       """
       print("Hello, World!")

   # Accessing the docstring
   print(example_function.__doc__)
   ```

## 16. Variable Naming

#### Variable Naming in Python

When naming variables in Python, there are specific rules and best practices to follow to ensure that the code is readable and maintainable. Here are the key points:

#### 1. Rules for Naming Variables
- **Alphanumeric and Underscores:** Variable names can contain letters (a-z, A-Z), digits (0-9), and underscores (_).
  ```python
  my_variable = 10
  variable1 = 20
  _variable = 30
  ```

- **Cannot Start with a Digit:** Variable names must begin with a letter or an underscore, but not a digit.
  ```python
  valid_variable = 10
  _valid_variable = 20
  # 1invalid_variable = 30  # Invalid, will cause a syntax error
  ```

- **Case Sensitivity:** Variable names are case-sensitive, meaning `myVariable` and `myvariable` are different variables.
  ```python
  myVariable = 10
  myvariable = 20
  ```

- **No Reserved Words:** Variable names cannot be Python reserved words (keywords) like `if`, `else`, `while`, etc.
  ```python
  # if = 10  # Invalid, will cause a syntax error
  my_if_variable = 10  # Valid
  ```

#### 2. Best Practices for Naming Variables
- **Descriptive Names:** Use descriptive names that convey the purpose of the variable.
  ```python
  counter = 10
  total_sum = 100
  ```

- **Use of Underscores:** Use underscores to separate words in variable names for better readability (snake_case).
  ```python
  student_name = "John"
  total_amount = 500
  ```

- **Avoid Single-letter Names:** Avoid using single-letter names except for counters or temporary variables.
  ```python
  for i in range(10):  # Acceptable for a loop counter
      print(i)
  ```

- **Consistent Naming Convention:** Stick to a consistent naming convention throughout the codebase.
  ```python
  # snake_case for variables and functions
  student_name = "Alice"
  calculate_total()

  # CamelCase for class names
  class MyClass:
      pass
  ```

#### Example of Good Variable Naming
```python
# Good variable names
student_name = "Alice"
total_score = 95.5
number_of_students = 30
is_valid = True

# Loop with a meaningful variable name
for student in students_list:
    print(student)
```

#### Example of Bad Variable Naming
```python
# Bad variable names
x = "Bob"
y = 85.5
z = 25

# Using single-letter names that don't convey meaning
for s in sl:
    print(s)
```

By following these rules and best practices, you can write code that is more readable, maintainable, and less prone to errors.
