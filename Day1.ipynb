### Types of Errors in Python

Errors in Python can be broadly categorized into three types: **Syntax Errors**, **Runtime Errors**, and **Logical Errors**.

---

#### **1. Syntax Errors**
Syntax errors occur when the Python parser encounters a syntactically incorrect statement. These errors are detected before the code is executed.

- Example:
  ```python
  print("Hello World"  # Missing closing parenthesis
  ```

---

#### **2. Runtime Errors**
Runtime errors occur while the program is being executed. They are exceptions raised when Python encounters something it cannot handle.

**Common types of runtime errors include:**

- **NameError**: Occurs when a variable or function name is not defined.
  ```python
  print(variable)  # 'variable' is not defined
  ```

- **TypeError**: Raised when an operation or function is applied to an object of inappropriate type.
  ```python
  result = "10" + 5  # Cannot add a string and an integer
  ```

- **IndexError**: Occurs when trying to access an index that is out of range for a list or other sequence types.
  ```python
  my_list = [1, 2, 3]
  print(my_list[5])  # Index out of range
  ```

- **ValueError**: Raised when a function receives an argument of the correct type but an invalid value.
  ```python
  number = int("hello")  # Cannot convert 'hello' to an integer
  ```

- **KeyError**: Occurs when trying to access a key that does not exist in a dictionary.
  ```python
  my_dict = {"a": 1}
  print(my_dict["b"])  # Key 'b' does not exist
  ```

- **AttributeError**: Raised when an attribute reference or assignment fails.
  ```python
  number = 10
  number.append(5)  # 'int' object has no attribute 'append'
  ```

- **ZeroDivisionError**: Raised when dividing a number by zero.
  ```python
  result = 10 / 0  # Division by zero
  ```

- **FileNotFoundError**: Occurs when trying to open a file that does not exist.
  ```python
  with open("non_existent_file.txt") as file:
      content = file.read()  # File not found
  ```

- **ImportError**: Raised when an import statement fails to find the module or a specific function/class in a module.
  ```python
  import nonexistent_module  # Module does not exist
  ```

- **IndentationError**: Occurs when the indentation is not consistent.
  ```python
  def my_function():
  print("Error")  # Indentation is incorrect
  ```

---

#### **3. Logical Errors**
Logical errors occur when the code runs without any syntax or runtime errors but produces incorrect results due to flawed logic.

- Example:
  ```python
  def add_numbers(a, b):
      return a - b  # Incorrect logic for addition

  print(add_numbers(10, 5))  # Output will be 5 instead of 15
  ```

### Types of Comments in Python

1. **Single-line Comments**
   - Start with `#` and extend to the end of the line.
   ```python
   # This is a single-line comment
   x = 5  # Inline comment
   ```

2. **Multi-line Comments**
   - Use multiple `#` characters or triple-quoted strings (`"""` or `'''`).
   ```python
   # This is a multi-line comment
   # that spans multiple lines.

   """
   This is a multi-line comment
   using triple-quoted strings.
   """
   ```

3. **Docstrings**
   - Used to document functions, classes, or modules. Written using triple-quoted strings and accessible via the `__doc__` attribute.
   ```python
   def example_function():
       """
       This is a docstring.
       It explains the purpose of the function.
       """
       print("Hello, World!")

   # Accessing the docstring
   print(example_function.__doc__)
   ```

- Ctrl + / = comment
- ctrl + z = undo

````python
def example_function():
    """
    This is a docstring.
    It explains the purpose of the function.
    """
    print("Hello, World!")

# Accessing the docstring
print(example_function.__doc__)
````

## 16. Variable Naming

#### Variable Naming in Python

1. **Must Begin with a Letter or an Underscore (`_`):**  
   Variable names must start with an alphabet (`a-z`, `A-Z`) or an underscore (`_`). They cannot start with a number.  
   - ✅ Valid: `my_var`, `_temp`, `data1`  
   - ❌ Invalid: `1data`, `@value`

---

2. **Can Contain Letters, Numbers, and Underscores:**  
   After the first character, you can use letters (`a-z`, `A-Z`), numbers (`0-9`), and underscores (`_`).  
   - ✅ Valid: `my_var2`, `result_value`, `data_123`  
   - ❌ Invalid: `data-value`, `first$name`

---

3. **Cannot Use Reserved Keywords:**  
   Variable names cannot be any of Python's reserved keywords (e.g., `if`, `else`, `while`, `True`, `None`, etc.).  
   - ❌ Invalid: `if = 5`, `class = 10`  
   Use descriptive names instead:
   - ✅ Valid: `if_condition`, `class_data`

---

4. **Case-Sensitive:**  
   Variable names are case-sensitive. For example, `Variable`, `variable`, and `VARIABLE` are considered different.  
   - Example:
     ```python
     age = 25
     Age = 30
     print(age)  # Output: 25
     print(Age)  # Output: 30
     ```

---

5. **No Spaces Allowed:**  
   Variable names cannot contain spaces. Use underscores (`_`) to separate words for better readability.  
   - ✅ Valid: `user_name`, `total_count`  
   - ❌ Invalid: `user name`, `total count`

---

6. **Avoid Using Special Characters:**  
   Only underscores (`_`) are allowed as a special character. Avoid using symbols like `@`, `$`, `!`, etc.  
   - ✅ Valid: `_temp`, `variable_1`  
   - ❌ Invalid: `temp@`, `var#name`

---

7. **Should Not Begin with a Number:**  
   Variable names cannot start with a number, though numbers can appear after the first letter or underscore.  
   - ✅ Valid: `var1`, `_1name`  
   - ❌ Invalid: `1variable`, `123name`

---

8. **Follow Descriptive and Readable Naming Practices (Recommended):**  
   Use meaningful names that indicate the purpose of the variable. This is not a rule but a good practice.  
   - ✅ Good: `student_name`, `total_score`, `is_logged_in`  
   - ❌ Bad: `x`, `y`, `a1`

---

#### Example of Good Variable Naming
```python
# Good variable names
student_name = "Alice"
total_score = 95.5
number_of_students = 30
is_valid = True

# Loop with a meaningful variable name
for student in students_list:
    print(student)
```

#### Example of Bad Variable Naming
```python
# Bad variable names
x = "Bob"
y = 85.5
z = 25

# Using single-letter names that don't convey meaning
for s in sl:
    print(s)
```
---

### Example of Valid and Invalid Variable Names:

| **Valid**           | **Invalid**      |
|----------------------|------------------|
| `user_name`         | `user-name`      |
| `_temp`             | `2temp`          |
| `value1`            | `value#1`        |
| `total_score`       | `total score`    |
| `data_set`          | `class` (keyword)|

By following these rules, you can create clear, error-free variable names in Python!
