### Indentation 

It is crucial for defining code blocks, such as loops and conditionals. It groups statements together and determines the structure and flow of your program. 

- **What is it?** Indentation refers to the spaces at the beginning of a line of code used to group statements.
- **Why is it important?** Python uses indentation instead of braces or keywords to define code blocks.
- **How many spaces?** The standard is 4 spaces per level, but consistency is key.
- **Mixing tabs and spaces?** Avoid mixing them to prevent debugging issues.

**Example:**


## @ Nested if statements and elif statements

```python
if condition:
    if another_condition:
        # do this
    else:
        # do this
else:
    # do this
```

Here's a quick breakdown:
- The first `if` checks the `condition`.
- If `condition` is `True`, it will then check `another_condition`.
  - If `another_condition` is `True`, it executes the code block under the inner `if`.
  - If `another_condition` is `False`, it executes the code block under the inner `else`.
- If the first `if` condition is `False`, it skips to the code block under the outer `else`.

````python
age = int(input("Enter your age: "))
if(age<=18):
    print("You have to pay $7.")
else:
    print("You have to pay $12.")
````
